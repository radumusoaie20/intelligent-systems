# LLMs

    LLM-urile, abreviat de la Large Language Models, sunt modele predictive de propoziții care folosesc metode de AI pentru a putea
    menține o conversație cu un utilizator (prin metoda Transformer, unde există mai multe blocuri succesive de rețele neuronale și blocuri de atenție)

# Agent

    Un agent reprezintă o entitate menită să interacționeze cu utilizatorul prin cadrul tool-urilor pe care le deține. Adesea, un tool dețiunut
    de agent este un LLM, pentru a putea trimite răspunsuri utilizatorului.

# CrewAI

    Crewai este o librărie de Python menită pentru creearea de agenți și sarcinii (preluate și executate) de agenți, și orchestrarea agențiilor pentru a realiza sarciniile.

# Ollama

     Ollama este un tool care poate să ruleze modele LLM local pe calculatorul tău. Servește un api de tip REST la care framework-uri precum CrewAI pot apela
     server-ul de Ollama pentru a putea trimite un prompt unui model LLM.

# Instalare Ollama



Se deschide un terminal care prezintă `curl` și suport pentru `bash` (pe Windows se poate folosi Git Bash) se execută următoarele comenzi (pas cu pas)

- `curl -fsSL https://ollama.com/install.sh | sh` (instalare ollama)
- `ollama serve` (pornire server)
- `ollama pull gemma3:270m` (modelul mic)
- `ollama pull gemma3:1b` (modelul mai mare)
- `ollama list` (verificare instalare model)
